```python
from sklearn.linear_model import SGDClassifier
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Sample dataset (inputs and responses)
training_data = [
    ("Hi", "Hello! How can I assist you today?"),
    ("Hello", "Hi! What can I do for you?"),
    ("How are you?", "I'm a chatbot, so I don't have feelings, but thanks for asking!"),
    ("What's your name?", "I'm your chatbot assistant!"),
    ("Bye", "Goodbye! Have a great day!"),
    ("See you", "Take care! See you soon!"),
    ("When did the Postulate start?", "2013"),
    ("Who is the originator of the Postulate?", "Dr. Johnson Durai Raj"),
    ("Is your company ISO certified?", "Yes, Postulate is the ISO Certified Quality Standard Company"),
    ("Where is the Postulate's headquarters located?", "Thoothukudi"),
    ("In what field is the Postulate specified?", "AI and Software development"),
    ("What is the vision of the Postulate?", "Postulate aims to enrich human society by igniting young minds. It imparts and transforms knowledge on recent technologies."),
    ("What services does your company provide?", "Data Analytics, Software Development, Web Development, Mobile Application"),
    ("What courses did Postulate provide?", "Data Science and Artificial Intelligence, Full Stack Development, Core Programming, Core Development, Game Development, Big Data"),
    ("What is the fee structure for Data Science and Artificial Intelligence?", "28,000"),
    ("Is there any discount for courses at Postulate?", "Yes, we offer 5% for female candidates and 10% for full payment")
]




# Split data into inputs and responses
inputs = [item[0] for item in training_data]
responses = [item[1] for item in training_data]

# Label encoding for responses
le = LabelEncoder()
encoded_responses = le.fit_transform(responses)

# Create the SLM pipeline: vectorizer + SVM classifier
model = make_pipeline(
    CountVectorizer(),  # Converts text into word count vectors
    SGDClassifier(loss="log_loss")  # SVM-based classifier with logistic loss (SLM-like approach)
)

# Train the model
model.fit(inputs, encoded_responses)

# Function to make chatbot responses
def chatbot_response(user_input):
    try:
        # Predict the response category
        response_code = model.predict([user_input])[0]
        # Decode the response code to text
        response = le.inverse_transform([response_code])[0]
    except Exception as e:
        response = "Sorry, I don't understand that."
    return response

# Chat with the bot
def chat():
    print("Chatbot: Hi! How can I help you? (Type 'exit' to stop)")
    while True:
        user_input = input("You: ")
        if user_input.lower() == "exit":
            print("Chatbot: Goodbye!")
            break
        bot_response = chatbot_response(user_input)
        print(f"Chatbot: {bot_response}")

if __name__ == "__main__":
    chat()
```
